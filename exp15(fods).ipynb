```python
#importing Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#load dataset
df=pd.read_csv('/content/healthcare-dataset-stroke-data.csv')
#display first 10 rows
print("First 10 rows of dataset:- ")
print(df.head(10))
```

```python
#Basic info
print("\nBasic Info:-")
print(df.info())
```

```python
#check for missing values
print("\nMissing values:-")
print(df.isnull().sum())
```

```python
#drop duplicates
df= df.drop_duplicates()
```

```python
#Basic Statistics
print("\nBasic Statistics:-")
print(df.describe())
```

```python
#data Cleaning
df.columns = df.columns.str.strip()
df = df.dropna()
#correlation for numeric features
corr = df.corr(numeric_only=True)
```

```python
#Data visulazation
#Distribution of BMI(Body Mass Index)
plt.figure(figsize=(5,4))
sns.histplot(df['bmi'], bins=30)
plt.title('Distribution of BMI')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

```python
#Bar Chart for Gender
gender_count= df['gender'].value_counts()
plt.figure(figsize=(5,4))
plt.bar(gender_count.index, gender_count.values,color=['skyblue','pink'])
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
#Scatter plot: BMI VS average glucose level
plt.figure(figsize=(5,4))
sns.scatterplot(x='bmi', y='avg_glucose_level', data=df,hue='gender')
plt.title('BMI vs Average Glucose Level')
plt.xlabel('BMI')
plt.ylabel('Average Glucose Level')
```

```python
#Box plot:-BMI by Smoking Status
plt.figure(figsize=(10,6))
sns.boxplot(x='smoking_status', y='bmi', data=df,hue='smoking_status',palette='Set2',legend=False)
plt.xticks(rotation=45)
plt.title('BMI by Smoking Status')
plt.xlabel('Smoking Status')
plt.ylabel('BMI')
plt.show()
```

```python
#Line Plot: Average Glucose Level by Age
avg_glucose_by_age = df.groupby('age')['avg_glucose_level'].mean()
plt.figure(figsize=(10,6))
plt.plot(avg_glucose_by_age.index, avg_glucose_by_age.values, marker='o', linestyle='-')
plt.title('Average Glucose Level by Age')
plt.xlabel('Age')
plt.ylabel('Average Glucose Level')
plt.grid(True)
plt.show()
```

```python
#Pairplot
sns.pairplot(df,vars=['age','avg_glucose_level', 'bmi'],hue='gender')
plt.show()
```
